```python
# Create a copy of the original dataset

clean_df = df.copy()

print("Original Dataset Shape:", clean_df.shape)
```

```python
# Remove missing and zero ratings

clean_df = clean_df.dropna(subset=['Aggregate rating'])

clean_df = clean_df[clean_df['Aggregate rating'] > 0]

print("Unrated restaurants removed!")
print("New Dataset Shape:", clean_df.shape)
```

```python
# Select input features

features = [
    'Votes',
    'Average Cost for two',
    'Price range',
    'Has Online delivery',
    'Has Table booking',
    'City',
    'Cuisines'
]

X = clean_df[features]

# Target variable

y = clean_df['Aggregate rating']

print("Features selected successfully!")

print("Input Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
# Convert categorical columns into numerical columns

X_encoded = pd.get_dummies(
    X,
    columns=['Has Online delivery',
             'Has Table booking',
             'City',
             'Cuisines'],
    drop_first=True
)

print("Categorical data converted successfully!")

print("Encoded Dataset Shape:", X_encoded.shape)

X_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42
)

print("Data splitting completed!")

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
print("Training Features:")
print(X_train.head())

print("\nTraining Target:")
print(y_train.head())

print("\nTesting Features Shape:", X_test.shape)
print("Testing Target Shape:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

print("Machine Learning algorithms imported successfully!")
```

```python
# Create model

lr_model = LinearRegression()

# Train model

lr_model.fit(X_train, y_train)

# Make predictions

lr_predictions = lr_model.predict(X_test)

print("Linear Regression model trained successfully!")
```

```python
dt_model = DecisionTreeRegressor(
    random_state=42,
    max_depth=10
)

dt_model.fit(X_train, y_train)

dt_predictions = dt_model.predict(X_test)

print("Decision Tree model trained successfully!")
```

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    max_depth=15,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)

print("Random Forest model trained successfully!")
```

```python
# Compare actual and predicted ratings

comparison = pd.DataFrame({
    'Actual Rating': y_test.values,
    'Linear Regression': lr_predictions,
    'Decision Tree': dt_predictions,
    'Random Forest': rf_predictions
})

print(comparison.head(10))
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("Evaluation metrics imported successfully!")
```

```python
import numpy as np

models = {
    "Linear Regression": lr_predictions,
    "Decision Tree": dt_predictions,
    "Random Forest": rf_predictions
}

results = []

for name, predictions in models.items():

    mae = mean_absolute_error(y_test, predictions)

    mse = mean_squared_error(y_test, predictions)

    rmse = np.sqrt(mse)

    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    })

results_df = pd.DataFrame(results)

print(results_df.round(4))
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=results_df,
    x="Model",
    y="R2 Score"
)

plt.title("Comparison of Machine Learning Models")
plt.ylabel("R2 Score")
plt.xlabel("Machine Learning Models")

plt.show()
```

```python
best_model_name = results_df.loc[
    results_df["R2 Score"].idxmax(),
    "Model"
]

print("Best Performing Model:", best_model_name)
```

```python
# Select the best model based on R2 Score

best_model_name = results_df.loc[
    results_df["R2 Score"].idxmax(),
    "Model"
]

if best_model_name == "Linear Regression":
    best_model = lr_model

elif best_model_name == "Decision Tree":
    best_model = dt_model

else:
    best_model = rf_model

print("Selected Model:", best_model_name)
```

```python
# Create sample restaurant data

new_restaurant = pd.DataFrame([{
    'Votes': 500,
    'Average Cost for two': 800,
    'Price range': 2,
    'Has Online delivery': 'Yes',
    'Has Table booking': 'No',
    'City': clean_df['City'].mode()[0],
    'Cuisines': clean_df['Cuisines'].mode()[0]
}])

# Encode categorical features

new_restaurant_encoded = pd.get_dummies(
    new_restaurant,
    columns=[
        'Has Online delivery',
        'Has Table booking',
        'City',
        'Cuisines'
    ],
    drop_first=True
)

# Match training columns

new_restaurant_encoded = new_restaurant_encoded.reindex(
    columns=X_train.columns,
    fill_value=0
)

# Predict rating

predicted_rating = best_model.predict(
    new_restaurant_encoded
)[0]

print("Predicted Restaurant Rating:", round(predicted_rating, 2))
```
